## Hiperparámetros usados
- Discretización: n_bins = 20 (cuadrícula 20×20 = 400 estados posibles)
- Tasa de aprendizaje (alpha / lr): 0.1
- Factor de descuento (gamma): 0.99
- Epsilon: inicial 1.0 → final 0.01, decaimiento 0.9995 por episodio
- Episodios de entrenamiento: 10.000

## Resultado de la evaluación (10 episodios, modo determinista)
- Recompensa promedio: **-149.30 ± 15.99**
- Llegó a la bandera: **10/10 episodios**
- Estados visitados: 294 / 400

## Comentario
El agente aprendió a completar el entorno de manera consistente, alcanzando la bandera en las 10 evaluaciones. Durante el entrenamiento, la recompensa promedio mejoró de -200 (inicio con política aleatoria) a cerca de -130, y la política final se situó en torno a -149 durante la evaluación. La discretización 20×20 resultó suficiente para representar la dinámica de posición y velocidad. La ligera variabilidad (±15.99) es normal en un método tabular con exploración residual (epsilon = 0.01) y una cuadrícula bastante gruesa.
